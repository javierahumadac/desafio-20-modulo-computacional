```python
# librerias
import pandas as pd
```

```python
# ubicar y abrir archivos

estudiantes_postulaciones = "C:/Users/salb7/OneDrive/Escritorio/U/USM 2024-02/MEMORIA/Base de datos - Estudiantes.csv"
desafios = "C:/Users/salb7/OneDrive/Escritorio/U/USM 2024-02/MEMORIA/Base de datos - Desafios.csv"

datos_estudiantes = pd.read_csv(estudiantes_postulaciones)
datos_desafios = pd.read_csv(desafios)

print(datos_estudiantes.columns)
print(datos_desafios.columns)
```

```python
# datos
# estudiantes
nombres_estudiantes = datos_estudiantes['Unnamed: 0'].tolist()
carrera_estudiantes = datos_estudiantes['Unnamed: 1'].tolist()
# desafios
nombre_desafios = datos_desafios['Nombre'].tolist()
carreras_afin_desafio = datos_desafios['Carreras'].tolist()
```

```python
# ids (estudiantes y desafios)

id_estudiantes = {nombre: idx for idx, nombre in enumerate(datos_estudiantes['Unnamed: 0'].unique())}
id_desafios = {nombre_desafio: idx for idx, nombre_desafio in enumerate(datos_desafios['Nombre'].unique())}

datos_estudiantes['ID Estudiante'] = datos_estudiantes['Unnamed: 0'].map(id_estudiantes)
datos_estudiantes['postulaciones_completas'] = datos_estudiantes.apply(lambda row: [row['Unnamed: 2'], row['Unnamed: 3'], row['Unnamed: 4']], axis=1)
# Modificar solo si el nombre de desafío completo debe ser mapeado
datos_desafios['ID desafio'] = datos_desafios['Nombre'].apply(lambda x: id_desafios.get(x, None))
```

```python
# Convertir las listas de postulaciones a una cadena, ignorando valores NaN o None
datos_estudiantes['postulaciones_como_cadena'] = datos_estudiantes['postulaciones_completas'].apply(
    lambda lista: '; '.join([str(item) for item in lista if pd.notna(item)]) if isinstance(lista, list) else None
)
```
